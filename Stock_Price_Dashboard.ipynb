```python
import streamlit as st
import yfinance as yf
import pandas as pd
import plotly.graph_objs as go
from datetime import datetime,timedelta
```

```python
st.sidebar.title('Stock Price Dashboard')
ticker = st.sidebar.text_input('Enter Stock Ticker','NVDA')
start_date =st.sidebar.date_input('Start Date',datetime.today()-timedelta(days=365))
end_date = st.sidebar.date_input('End Date',datetime.today())
```

```python
data =yf.download(ticker,start=start_date,end=end_date)
```

```python
st.title(f"{ticker}'Stock Price Dashboard")
```

```python
#Visual
st.subheader('Closing Price')
fig =go.Figure()
fig.add_trace(go.Scatter(x=data.index,y=data['Close'],mode='lines',name='Çlose'))
fig.update(layout_xaxis_rangeslider_visible=True)
st.plotly_chart(fig)
```

```python
#Volume chart
st.subheader('Volume')
vol_fig =go.Figure()
vol_fig.add_trace(go.Scatter(x=data.index,y=data['Volume'],name='Volume'))
vol_fig.update(layout_xaxis_rangeslider_visible=True)
st.plotly_chart(vol_fig)
```

```python
#Moving average
st.subheader('Moving Average')
mav_fig =go.Figure()
mav_fig.add_trace(go.Scatter(x=data.index,y=data['Close'],mode='lines',name='Close'))
mav_fig.add_trace(go.Scatter(x=data.index,y=data['MAV_20'],mode='lines',name='Moving Avg 20'))
mav_fig.add_trace(go.Scatter(x=data.index,y=data['MAV_50'],mode='lines',name='Moving Avg 50'))
mav_fig.update(layout_xaxis_rangeslider_visible=True)
st.plotly_chart(mav_fig)
```

```python
data['MAV_20'] = data['Close'].rolling(window=20).mean()
data['MAV_50'] = data['Close'].rolling(window=50).mean()
```

```python
st.subheader('Raw Data')
st.write(data)
```

```python
if data.empty:
    st.warning("No data found. Please check the ticker or date range.")
else:
    st.subheader('Raw Data')
    st.write(data)
```
